# Projeto: Cálculo de Métricas de Avaliação

Este notebook tem como objetivo calcular as principais métricas de avaliação para modelos de classificação de dados.

### Matriz de Confusão

As métricas são baseadas nos seguintes valores de uma matriz de confusão:
- **VP (Verdadeiro Positivo):** O modelo previu "Positivo" e o valor real era "Positivo". (Acerto)
- **VN (Verdadeiro Negativo):** O modelo previu "Negativo" e o valor real era "Negativo". (Acerto)
- **FP (Falso Positivo):** O modelo previu "Positivo", mas o valor real era "Negativo". (Erro Tipo I)
- **FN (Falso Negativo):** O modelo previu "Negativo", mas o valor real era "Positivo". (Erro Tipo II)

Na célula a seguir, você pode definir esses valores de forma arbitrária para testar os cálculos.

```python
# ===================================================================
# valores da Matriz de Confusão aqui
# ===================================================================
# Altere estes valores para testar diferentes cenários.
# Exemplo de uma matriz de confusão arbitrária:
#
#     Previsto: Sim | Previsto: Não
# Real: Sim |   VP=90     |   FN=20
# Real: Não |   FP=10     |   VN=80
# -------------------------------------------------------------------

VP = 90  # Verdadeiros Positivos
VN = 80  # Verdadeiros Negativos
FP = 10  # Falsos Positivos (Erro Tipo I)
FN = 20  # Falsos Negativos (Erro Tipo II)

print("Valores de entrada definidos com sucesso:")
print(f"VP={VP}, VN={VN}, FP={FP}, FN={FN}")
```

```python
# ===================================================================
# Definição das Funções de Cálculo
# ===================================================================
# Esta célula contém as funções que implementam as fórmulas da Tabela 1.
# -------------------------------------------------------------------

def calcular_sensibilidade(vp, fn):
  """Calcula a Sensibilidade (Recall): VP / (VP + FN)"""
  if (vp + fn) == 0:
    return 0.0
  return vp / (vp + fn)

def calcular_especificidade(vn, fp):
  """Calcula a Especificidade: VN / (FP + VN)"""
  if (fp + vn) == 0:
    return 0.0
  return vn / (fp + vn)

def calcular_acuracia(vp, vn, fp, fn):
  """Calcula a Acurácia: (VP + VN) / (VP + VN + FP + FN)"""
  total_elementos = vp + vn + fp + fn
  if total_elementos == 0:
    return 0.0
  return (vp + vn) / total_elementos

def calcular_precisao(vp, fp):
  """Calcula a Precisão: VP / (VP + FP)"""
  if (vp + fp) == 0:
    return 0.0
  return vp / (vp + fp)

def calcular_f_score(precisao, sensibilidade):
  """Calcula o F-score: 2 * (Precisão * Sensibilidade) / (Precisão + Sensibilidade)"""
  if (precisao + sensibilidade) == 0:
    return 0.0
  return 2 * (precisao * sensibilidade) / (precisao + sensibilidade)

print("Funções para cálculo das métricas foram definidas.")
```

```python
# ===================================================================
# Execução dos Cálculos e Exibição dos Resultados
# ===================================================================
# Esta célula utiliza os valores da Célula 2 e as funções da Célula 3
# para calcular e mostrar os resultados finais.
# -------------------------------------------------------------------

# Chamada das funções para obter os resultados
sensibilidade = calcular_sensibilidade(VP, FN)
especificidade = calcular_especificidade(VN, FP)
acuracia = calcular_acuracia(VP, VN, FP, FN)
precisao = calcular_precisao(VP, FP)
f_score = calcular_f_score(precisao, sensibilidade)

# Exibição dos resultados
print("="*40)
print("  MÉTRICAS DE AVALIAÇÃO DO MODELO")
print("="*40)
print(f"Valores da Matriz de Confusão:")
print(f"VP: {VP} | FN: {FN}")
print(f"FP: {FP} | VN: {VN}\n")
print(f"Total de Elementos (N): {VP + VN + FP + FN}")
print("-"*40)
# A formatação ':.2%'} converte o valor para porcentagem com 2 casas decimais
print(f"Acurácia:               {acuracia:.2%}")
print(f"Sensibilidade (Recall): {sensibilidade:.2%}")
print(f"Especificidade:         {especificidade:.2%}")
print(f"Precisão:               {precisao:.2%}")
print(f"F-score:                {f_score:.4f}") # F-score é geralmente mostrado como decimal
print("="*40)
```
